```python
import matplotlib.pyplot as plt 
import numpy as np 
import pandas as pd 
import seaborn as sns
```

```python
path_vendas = "./datalake/clean_data/vendas.csv"
path_vendedores = "./datalake/clean_data/vendedores.csv"
path_clientes = "./datalake/clean_data/clientes.csv"
path_crediario = "./datalake/clean_data/crediario.csv"
path_promocoes = "./datalake/clean_data/promocoes.csv"
path_estoque = "./datalake/clean_data/estoque.csv"

df_vendas = pd.read_csv(path_vendas)
df_vendedores = pd.read_csv(path_vendedores)
df_clientes = pd.read_csv(path_clientes)
df_crediario = pd.read_csv(path_crediario)
df_promocoes = pd.read_csv(path_promocoes)
df_estoque = pd.read_csv(path_estoque)
```

```python
df_vendas['data_venda'] = pd.to_datetime(df_vendas['data_venda'])

vendas_mensais = df_vendas.resample('ME', on='data_venda')['valor_total'].sum().reset_index()

plt.figure(figsize=(12, 5))
sns.lineplot(data=vendas_mensais, x='data_venda', y='valor_total', marker='o', color='navy', linewidth=2.5)
plt.title('Evolução do Valor Total de Vendas por Período', fontsize=14, fontweight='bold')
plt.xlabel('Data (Mês)')
plt.ylabel('Valor Total (R$)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
vendas_por_categoria = df_vendas['categoria'].value_counts().reset_index()
vendas_por_categoria.columns = ['categoria', 'total_vendas']

plt.figure(figsize=(10, 5))
sns.barplot(data=vendas_por_categoria, y='categoria', x='total_vendas', palette='Blues_r')
plt.title('Número de Vendas por Categoria', fontsize=14, fontweight='bold')
plt.xlabel('Quantidade de Transações')
plt.ylabel('Categoria')
plt.tight_layout()
plt.show()
```

```python
qtd_produtos = df_vendas.groupby('produto')['quantidade'].sum().reset_index()
qtd_produtos = qtd_produtos.sort_values(by='quantidade', ascending=False)

plt.figure(figsize=(12, 6))
sns.barplot(data=qtd_produtos, x='produto', y='quantidade', palette='viridis')
plt.title('Produtos Vendidos (Quantidade Total)', fontsize=14, fontweight='bold')
plt.xlabel('Produto')
plt.ylabel('Quantidade de Unidades Vendidas')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
vendas_cliente = df_vendas.merge(df_clientes, on='id_cliente', how='inner')
pref_categoria = vendas_cliente.groupby(['canal_preferencial', 'categoria']).size().reset_index(name='contagem')

plt.figure(figsize=(12, 6))
sns.barplot(data=pref_categoria, x='categoria', y='contagem', hue='canal_preferencial', palette='Set2')
plt.title('Preferência de Categoria por Canal Preferencial do Cliente', fontsize=14, fontweight='bold')
plt.xlabel('Categoria do Produto')
plt.ylabel('Quantidade de Compras')
plt.legend(title='Canal Preferencial')
plt.tight_layout()
plt.show()
```

```python
gasto_cliente = df_vendas.groupby('id_cliente').agg(
    gasto_total=('valor_total', 'sum'),
    total_itens=('quantidade', 'sum')
).reset_index()

perfil_cliente = df_clientes.merge(gasto_cliente, on='id_cliente', how='inner')

plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=perfil_cliente, 
    x='idade', 
    y='gasto_total', 
    hue='canal_preferencial', 
    size='total_itens', 
    sizes=(30, 300), 
    alpha=0.7, 
    palette='deep'
)
plt.title('Perfil do Cliente: Idade vs. Valor Total Gasto', fontsize=14, fontweight='bold')
plt.xlabel('Idade do Cliente')
plt.ylabel('Gasto Total Acumulado (R$)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.boxplot(data=df_vendas, x='categoria', y='valor_total', ax=axes[0], palette='Pastel1')
axes[0].set_title('Distribuição e Dispersão de Valor de Venda por Categoria', fontweight='bold')
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=45)

sns.pointplot(data=df_vendas, x='categoria', y='valor_total', ci='sd', ax=axes[1], color='crimson', capsize=0.2)
axes[1].set_title('Média de Valor com Barra de Desvio Padrão (±1 SD)', fontweight='bold')
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=45)

plt.tight_layout()
plt.show()
```

```python
df_estoque['divergencia'] = df_estoque['estoque_fisico'] - df_estoque['estoque_atual']
estoque_inconsistente = df_estoque[df_estoque['divergencia'] != 0]

plt.figure(figsize=(10, 5))
sns.barplot(data=estoque_inconsistente, x='produto', y='divergencia', palette='coolwarm')
plt.axhline(0, color='black', linestyle='--')
plt.title('Inconsistência de Estoque: Diferença (Físico - Atual)', fontsize=14, fontweight='bold')
plt.xlabel('Produto')
plt.ylabel('Diferença de Unidades')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

vendas_sem_cliente = df_vendas['id_cliente'].isna().value_counts().reset_index()
vendas_sem_cliente.columns = ['Status_Cliente', 'Quantidade']
vendas_sem_cliente['Status_Cliente'] = vendas_sem_cliente['Status_Cliente'].map({True: 'Cliente Não Identificado (NaN)', False: 'Cliente Identificado'})

plt.figure(figsize=(7, 4))
sns.barplot(data=vendas_sem_cliente, x='Status_Cliente', y='Quantidade', palette='Set1')
plt.title('Inconsistência em Vendas: Registros Sem ID de Cliente', fontsize=12, fontweight='bold')
plt.ylabel('Total de Registros')
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 5))
sns.histplot(df_crediario['dias_atraso'], kde=True, color='darkorange', bins=30)
plt.axvline(0, color='red', linestyle='--', label='Vencimento (0 dias)')
plt.title('Distribuição dos Dias de Atraso das Parcelas do Crediário', fontsize=14, fontweight='bold')
plt.xlabel('Dias de Atraso')
plt.ylabel('Frequência de Parcelas')
plt.legend()
plt.tight_layout()
plt.show()
```
