```python
import pandas as pd
```

```python
# 1. Carregamento do arquivo de vendas do HDFS/Data Lake
df_vendas = pd.read_csv('../vendas.csv')
```

```python
# 2. Mapper: Mapeia cada registro gerando pares (Chave, 1)
def mapper(linhas):
    pares = []
    for categoria in linhas:
        if pd.notna(categoria):
            cat_limpa = str(categoria).strip()
            pares.append((cat_limpa, 1))
    return pares
```

```python
# 3. Shuffle: Agrupa as ocorrências por chave (categoria)
def shuffle(mapped_data):
    agrupado = {}
    for chave, valor in mapped_data:
        if chave not in agrupado:
            agrupado[chave] = []
        agrupado[chave].append(valor)
    return agrupado
```

```python
# 4. Reducer: Reduz as listas agrupadas efetuando o somatório final
def reducer(agrupado):
    return {chave: sum(valores) for chave, valores in agrupado.items()}
```

```python
# Execução do pipeline MapReduce
dados_entrada = df_vendas['categoria'].tolist()
mapeado = mapper(dados_entrada)
agrupado = shuffle(mapeado)
resultado_mapreduce = reducer(agrupado)
```

```python
print("[MapReduce] Total de itens vendidos por Categoria:")
print(resultado_mapreduce)
```
